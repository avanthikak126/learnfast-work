````python
import calendar
import datetime as dt


def next_month(d):
    year, month = (
        (d.year + 1, 1)
        if d.month == 12
        else (d.year, d.month + 1)
    )

    return dt.date(
        year,
        month,
        min(d.day, calendar.monthrange(year, month)[1])
    )


when = dt.date(2026, 1, 31)

steps = [when]

for _ in range(3):
    steps.append(next_month(steps[-1]))

print([d.isoformat() for d in steps])
````

````python
%%writefile mybudget/transaction.py
"""Transactions (lessons 2-3)."""

from __future__ import annotations

import calendar
import datetime as dt
from dataclasses import dataclass
from decimal import Decimal

FREQUENCIES = ("weekly", "monthly")


@dataclass(frozen=True)
class Transaction:
    date: dt.date
    amount: Decimal
    category: str
    description: str = ""

    def __post_init__(self) -> None:
        if not isinstance(self.date, dt.date):
            raise ValueError(f"date must be a date, got {self.date!r}")

        if not isinstance(self.amount, Decimal) or not self.amount.is_finite():
            raise ValueError(f"amount must be a finite Decimal, got {self.amount!r}")

        if self.amount == 0:
            raise ValueError("amount must not be zero")

        if not self.category.strip():
            raise ValueError("category must not be empty")

    @property
    def kind(self) -> str:
        return "income" if self.amount > 0 else "expense"

    def occurrences(self, until: dt.date | None = None) -> list[Transaction]:
        return [self]


@dataclass(frozen=True)
class RecurringTransaction(Transaction):
    frequency: str = "monthly"

    def __post_init__(self) -> None:
        super().__post_init__()

        if self.frequency not in FREQUENCIES:
            raise ValueError(
                f"frequency must be one of {FREQUENCIES}, got {self.frequency!r}"
            )

    def _nth(self, n: int) -> dt.date:
        if self.frequency == "weekly":
            return self.date + dt.timedelta(weeks=n)

        year_offset, month_index = divmod(self.date.month - 1 + n, 12)
        year = self.date.year + year_offset
        month = month_index + 1

        day = min(
            self.date.day,
            calendar.monthrange(year, month)[1]
        )

        return dt.date(year, month, day)

    def occurrences(self, until: dt.date | None = None) -> list[Transaction]:
        until = until or self.date

        out = []
        n = 0

        while (when := self._nth(n)) <= until:
            out.append(
                Transaction(
                    when,
                    self.amount,
                    self.category,
                    self.description
                )
            )
            n += 1

        return out
````

````python
%%writefile mybudget/account.py
"""The account (lessons 2-3)."""

from __future__ import annotations

import datetime as dt
from decimal import Decimal

from .transaction import Transaction


class Account:
    def __init__(self, name: str = "Main"):
        self.name = name
        self._transactions: list[Transaction] = []

    @property
    def balance(self) -> Decimal:
        return sum(
            (t.amount for t in self._transactions),
            Decimal(0)
        )

    @property
    def transactions(self) -> tuple[Transaction, ...]:
        return tuple(self._transactions)

    def add(
        self,
        transaction: Transaction,
        until: dt.date | None = None
    ) -> list[Transaction]:
        return self.add_many(transaction.occurrences(until))

    def add_many(self, transactions) -> list[Transaction]:
        batch = sorted(transactions, key=lambda t: t.date)
        self._transactions.extend(batch)
        return batch
````

````python
%%writefile mybudget/reports.py
"""Reports (lessons 3-4): abstraction and polymorphism."""

from __future__ import annotations
from abc import ABC, abstractmethod
from collections import defaultdict
from decimal import Decimal
from .account import Account


class Report(ABC):
    title = "Report"

    @abstractmethod
    def generate(self, account: Account) -> list[tuple[str, Decimal]]:
        pass

    def render(self, account: Account) -> str:
        rows = self.generate(account)
        lines = [self.title, "-" * len(self.title)]
        lines += [
            f"{label:<20}{amount:>14,.2f}"
            for label, amount in rows
        ]
        if not rows:
            lines.append("(no transactions)")
        return "\n".join(lines)


class MonthlyReport(Report):
    title = "Net by month"

    def generate(self, account: Account):
        totals = defaultdict(Decimal)
        for t in account.transactions:
            totals[t.date.strftime("%Y-%m")] += t.amount
        return sorted(totals.items())


class CategoryReport(Report):
    title = "Spending by category"

    def generate(self, account: Account):
        totals = defaultdict(Decimal)
        for t in account.transactions:
            if t.amount < 0:
                totals[t.category] += -t.amount
        return sorted(totals.items(), key=lambda row: row[1], reverse=True)
````

````python
class TopExpensesReport(Report):
    title = "Biggest expenses"

    def __init__(self, n=5):
        self.n = n

    def generate(self, account):
        spend = [
            (f"{t.date} {t.category}", -t.amount)
            for t in account.transactions
            if t.amount < 0
        ]
        return sorted(
            spend,
            key=lambda row: row[1],
            reverse=True
        )[:self.n]


for report in [MonthlyReport(), CategoryReport(), TopExpensesReport(3)]:
    print(report.render(account))
    print()
````

````python
%%writefile mybudget/exceptions.py
from decimal import Decimal


class BudgetError(Exception):
    """Base class for application errors."""


class InvalidTransactionError(BudgetError, ValueError):
    pass


class InsufficientFundsError(BudgetError):
    def __init__(self, balance: Decimal, needed: Decimal):
        self.balance = balance
        self.needed = needed
        super().__init__(
            f"balance {balance:,.2f} cannot cover an expense of {needed:,.2f}"
        )


class BudgetExceededError(BudgetError):
    def __init__(self, category: str, limit: Decimal, would_be: Decimal):
        self.category = category
        self.limit = limit
        self.would_be = would_be
        super().__init__(
            f"'{category}' budget is {limit:,.2f} this month; "
            f"this would make it {would_be:,.2f}"
        )


class StatementImportError(BudgetError):
    def __init__(self, message: str, line: int | None = None):
        self.line = line
        super().__init__(f"line {line}: {message}" if line else message)
````

````python
%%writefile mybudget/transaction.py
from __future__ import annotations

import calendar
import datetime as dt
from dataclasses import dataclass
from decimal import Decimal, InvalidOperation

from .exceptions import InvalidTransactionError

FREQUENCIES = ("weekly", "monthly")


@dataclass(frozen=True)
class Transaction:
    date: dt.date
    amount: Decimal
    category: str
    description: str = ""

    def __post_init__(self) -> None:
        if not isinstance(self.date, dt.date):
            raise InvalidTransactionError(f"date must be a date, got {self.date!r}")
        if not isinstance(self.amount, Decimal) or not self.amount.is_finite():
            raise InvalidTransactionError(
                f"amount must be a finite Decimal, got {self.amount!r}"
            )
        if self.amount == 0:
            raise InvalidTransactionError("amount must not be zero")
        if not self.category.strip():
            raise InvalidTransactionError("category must not be empty")

    @property
    def kind(self) -> str:
        return "income" if self.amount > 0 else "expense"

    def occurrences(self, until: dt.date | None = None) -> list[Transaction]:
        return [self]

    @classmethod
    def parse(cls, date: str, amount: str, category: str,
              description: str = "", **extra):
        try:
            parsed_date = dt.date.fromisoformat(date.strip())
        except ValueError:
            raise InvalidTransactionError(
                f"date must look like 2026-03-15, got {date!r}"
            ) from None

        try:
            parsed_amount = Decimal(amount.strip())
        except InvalidOperation:
            raise InvalidTransactionError(
                f"amount must be a number, got {amount!r}"
            ) from None

        return cls(
            parsed_date,
            parsed_amount,
            category.strip().lower(),
            description.strip(),
            **extra
        )


@dataclass(frozen=True)
class RecurringTransaction(Transaction):
    frequency: str = "monthly"

    def __post_init__(self) -> None:
        super().__post_init__()
        if self.frequency not in FREQUENCIES:
            raise InvalidTransactionError(
                f"frequency must be one of {FREQUENCIES}, got {self.frequency!r}"
            )

    def _nth(self, n: int) -> dt.date:
        if self.frequency == "weekly":
            return self.date + dt.timedelta(weeks=n)

        year_offset, month_index = divmod(self.date.month - 1 + n, 12)
        year, month = self.date.year + year_offset, month_index + 1
        day = min(self.date.day, calendar.monthrange(year, month)[1])
        return dt.date(year, month, day)

    def occurrences(self, until: dt.date | None = None) -> list[Transaction]:
        until = until or self.date
        out = []
        n = 0

        while (when := self._nth(n)) <= until:
            out.append(
                Transaction(when, self.amount, self.category, self.description)
            )
            n += 1

        return out
````

````python
%%writefile mybudget/account.py
from __future__ import annotations

import datetime as dt
from collections import defaultdict
from decimal import Decimal

from .exceptions import (
    BudgetExceededError,
    InsufficientFundsError,
    InvalidTransactionError,
)
from .transaction import Transaction


class Account:
    def __init__(self, name: str = "Main", allow_overdraft: bool = False):
        self.name = name
        self.allow_overdraft = allow_overdraft
        self._transactions: list[Transaction] = []
        self._budgets: dict[str, Decimal] = {}

    @property
    def balance(self) -> Decimal:
        return sum((t.amount for t in self._transactions), Decimal(0))

    @property
    def transactions(self) -> tuple[Transaction, ...]:
        return tuple(self._transactions)

    @property
    def budgets(self) -> dict[str, Decimal]:
        return dict(self._budgets)

    def spent(self, category: str, year: int, month: int) -> Decimal:
        return sum(
            (
                -t.amount for t in self._transactions
                if t.category == category
                and t.amount < 0
                and (t.date.year, t.date.month) == (year, month)
            ),
            Decimal(0),
        )

    def set_budget(self, category: str, monthly_limit: Decimal) -> None:
        if monthly_limit <= 0:
            raise InvalidTransactionError("a budget limit must be positive")
        self._budgets[category.strip().lower()] = monthly_limit

    def add(self, transaction: Transaction,
            until: dt.date | None = None) -> list[Transaction]:
        return self.add_many(transaction.occurrences(until))

    def add_many(self, transactions,
                 *, enforce_limits: bool = True) -> list[Transaction]:
        batch = sorted(transactions, key=lambda t: t.date)

        if enforce_limits:
            self._check(batch)

        self._transactions.extend(batch)
        return batch

    def _check(self, batch: list[Transaction]) -> None:
        balance = self.balance
        spent = defaultdict(Decimal)

        for t in batch:
            balance += t.amount

            if t.amount > 0:
                continue

            if not self.allow_overdraft and balance < 0:
                raise InsufficientFundsError(balance - t.amount, -t.amount)

            limit = self._budgets.get(t.category)

            if limit is not None:
                key = (t.category, t.date.year, t.date.month)

                if key not in spent:
                    spent[key] = self.spent(*key)

                spent[key] += -t.amount

                if spent[key] > limit:
                    raise BudgetExceededError(t.category, limit, spent[key])
````

````python

%%writefile mybudget/storage.py
from __future__ import annotations

import json
import os
from dataclasses import dataclass, field
from decimal import Decimal, InvalidOperation
from pathlib import Path

from .account import Account
from .exceptions import BudgetError
from .transaction import Transaction

SCHEMA_VERSION = 1


@dataclass
class LoadResult:
    account: Account
    problems: list[str] = field(default_factory=list)


def save(account: Account, path: str | Path) -> None:
    path = Path(path)

    payload = {
        "version": SCHEMA_VERSION,
        "name": account.name,
        "allow_overdraft": account.allow_overdraft,
        "budgets": {
            category: str(value)
            for category, value in account.budgets.items()
        },
        "transactions": [
            {
                "date": t.date.isoformat(),
                "amount": str(t.amount),
                "category": t.category,
                "description": t.description,
            }
            for t in account.transactions
        ],
    }

    tmp = path.with_name(path.name + ".tmp")
    tmp.write_text(json.dumps(payload, indent=2), encoding="utf-8")
    os.replace(tmp, path)


def load(path: str | Path) -> LoadResult:
    path = Path(path)

    if not path.exists():
        return LoadResult(Account())

    try:
        data = json.loads(path.read_text(encoding="utf-8"))
        if not isinstance(data, dict):
            raise ValueError("top level is not an object")

    except (ValueError, UnicodeDecodeError) as exc:
        quarantine = path.with_name(path.name + ".corrupt")
        os.replace(path, quarantine)

        return LoadResult(
            Account(),
            [f"{path.name} was unreadable ({exc}); kept as {quarantine.name}"]
        )

    account = Account(
        name=str(data.get("name", "Main")),
        allow_overdraft=bool(data.get("allow_overdraft", False)),
    )

    problems = []

    for category, limit in (data.get("budgets") or {}).items():
        try:
            account.set_budget(category, Decimal(limit))
        except (InvalidOperation, TypeError, BudgetError) as exc:
            problems.append(f"budget {category!r} skipped: {exc}")

    good = []

    for index, record in enumerate(data.get("transactions") or []):
        try:
            good.append(
                Transaction.parse(
                    record["date"],
                    record["amount"],
                    record["category"],
                    record.get("description", ""),
                )
            )
        except (KeyError, TypeError, AttributeError, BudgetError) as exc:
            problems.append(
                f"transaction #{index} skipped: {exc!r}"
            )

    account.add_many(good, enforce_limits=False)

    return LoadResult(account, problems)
````

````python
%%writefile mybudget/importer.py

from __future__ import annotations

import csv
import time
from collections.abc import Callable

from .exceptions import BudgetError, StatementImportError
from .transaction import Transaction

REQUIRED_COLUMNS = ("date", "description", "amount", "category")
BATCH = 500


def parse_statement(
    text: str,
    on_progress: Callable[[float], None] | None = None,
    throttle: float = 0.0,
) -> list[Transaction]:

    lines = text.splitlines()

    if not lines:
        raise StatementImportError("the file is empty")

    reader = csv.DictReader(lines)

    missing = [
        c for c in REQUIRED_COLUMNS
        if c not in (reader.fieldnames or [])
    ]

    if missing:
        raise StatementImportError(
            f"missing column(s): {', '.join(missing)}",
            line=1
        )

    total = max(len(lines) - 1, 1)
    out = []

    for done, row in enumerate(reader, start=1):
        line_no = reader.line_num

        if None in row or None in row.values():
            raise StatementImportError(
                "wrong number of columns",
                line=line_no
            )

        try:
            out.append(
                Transaction.parse(
                    row["date"],
                    row["amount"],
                    row["category"],
                    row["description"]
                )
            )
        except BudgetError as exc:
            raise StatementImportError(
                str(exc),
                line=line_no
            ) from exc

        if done % BATCH == 0:
            if on_progress:
                on_progress(done / total)

            if throttle:
                time.sleep(throttle)

    if on_progress:
        on_progress(1.0)

    return out
````

````python
class StopException(BaseException):
    pass


class FakeSessionState:
    def __init__(self):
        self.values = {"status": "running", "progress": 0.0}
        self.current_run = 1

    def rerun(self):
        self.current_run += 1

    def write(self, run_id, key, value):
        if run_id != self.current_run:
            raise StopException()
        self.values[key] = value


def obvious_worker(session, run_id, text):
    try:
        parse_statement(
            text,
            lambda p: session.write(run_id, "progress", p),
            throttle=0.02
        )
        session.write(run_id, "status", "done")
    except Exception as exc:
        session.values["status"] = f"failed: {exc}"


def where_the_error_goes(args):
    print(
        f"[terminal] Exception in thread: "
        f"{args.exc_type.__name__}"
    )


original_hook = threading.excepthook
threading.excepthook = where_the_error_goes

session = FakeSessionState()

thread = threading.Thread(
    target=obvious_worker,
    args=(session, 1, big),
    daemon=True
)

thread.start()
time.sleep(0.3)
session.rerun()
time.sleep(1)

threading.excepthook = original_hook

print("Worker alive:", thread.is_alive())
print("Page sees:", session.values)
````

````python
%%writefile mybudget/importer.py

from __future__ import annotations

import csv
import threading
import time
from collections.abc import Callable

from .exceptions import BudgetError, StatementImportError
from .transaction import Transaction

REQUIRED_COLUMNS = ("date", "description", "amount", "category")
BATCH = 500


def parse_statement(text, on_progress=None, throttle=0.0):
    lines = text.splitlines()

    if not lines:
        raise StatementImportError("the file is empty")

    reader = csv.DictReader(lines)

    missing = [
        c for c in REQUIRED_COLUMNS
        if c not in (reader.fieldnames or [])
    ]

    if missing:
        raise StatementImportError(
            f"missing column(s): {', '.join(missing)}",
            line=1
        )

    total = max(len(lines) - 1, 1)
    out = []

    for done, row in enumerate(reader, start=1):
        line_no = reader.line_num

        if None in row or None in row.values():
            raise StatementImportError(
                "wrong number of columns",
                line=line_no
            )

        try:
            out.append(Transaction.parse(
                row["date"],
                row["amount"],
                row["category"],
                row["description"]
            ))
        except BudgetError as exc:
            raise StatementImportError(
                str(exc), line=line_no
            ) from exc

        if done % BATCH == 0:
            if on_progress:
                on_progress(done / total)
            if throttle:
                time.sleep(throttle)

    if on_progress:
        on_progress(1.0)

    return out


class ImportJob:

    def __init__(self):
        self._lock = threading.Lock()
        self._thread = None
        self._reset()

    def _reset(self):
        self._status = "idle"
        self._progress = 0.0
        self._result = None
        self._error = None

    @property
    def status(self):
        with self._lock:
            return self._status

    @property
    def progress(self):
        with self._lock:
            return self._progress

    @property
    def result(self):
        with self._lock:
            return self._result

    @property
    def error(self):
        with self._lock:
            return self._error

    def reset(self):
        with self._lock:
            self._reset()

    def start(self, text, throttle=0.0):
        with self._lock:
            if self._status == "running":
                raise RuntimeError("an import is already running")

            self._reset()
            self._status = "running"

        self._thread = threading.Thread(
            target=self._run,
            args=(text, throttle),
            daemon=True
        )

        self._thread.start()
        return self._thread

    def join(self, timeout=None):
        if self._thread is not None:
            self._thread.join(timeout)

    def _set_progress(self, value):
        with self._lock:
            self._progress = value

    def _run(self, text, throttle):
        try:
            result = parse_statement(
                text,
                self._set_progress,
                throttle
            )

            with self._lock:
                self._result = result
                self._status = "done"

        except Exception as exc:
            with self._lock:
                self._error = f"{type(exc).__name__}: {exc}"
                self._status = "failed"

        finally:
            with self._lock:
                if self._status == "running":
                    self._error = "the import worker stopped unexpectedly"
                    self._status = "failed"
````

````python
%%writefile test_mybudget.py
import datetime as dt
from decimal import Decimal

import pytest

from mybudget import (
    Account, BudgetExceededError, InsufficientFundsError,
    InvalidTransactionError, RecurringTransaction, Report,
    StatementImportError, Transaction, load, save,
)
from mybudget.importer import ImportJob, parse_statement


def tx(day, amount, category="misc"):
    return Transaction(
        dt.date.fromisoformat(day), Decimal(amount), category
    )


@pytest.mark.parametrize("amount", ["0", "NaN"])
def test_unusable_amounts_are_refused(amount):
    with pytest.raises(InvalidTransactionError):
        tx("2026-01-01", amount)


def test_money_adds_exactly():
    account = Account()
    for _ in range(1000):
        account.add(tx("2026-01-01", "0.10"))
    assert account.balance == Decimal("100.00")


def test_overdraft_is_refused():
    account = Account()
    account.add(tx("2026-01-01", "100"))

    with pytest.raises(InsufficientFundsError):
        account.add(tx("2026-01-02", "-150"))

    assert account.balance == Decimal("100")


def test_batch_is_all_or_nothing():
    account = Account()
    account.add(tx("2026-01-01", "100"))
    batch = [tx(f"2026-01-0{d}", "-30") for d in (2, 3, 4, 5)]

    with pytest.raises(InsufficientFundsError):
        account.add_many(batch)

    assert len(account.transactions) == 1


def test_budget_limit():
    account = Account()
    account.add(tx("2026-01-01", "1000"))
    account.set_budget("dining", Decimal("100"))
    account.add(tx("2026-01-05", "-60", "dining"))

    with pytest.raises(BudgetExceededError):
        account.add(tx("2026-01-06", "-50", "dining"))


def test_monthly_recurring_transaction():
    bill = RecurringTransaction(
        dt.date(2026, 1, 31), Decimal("-20"), "phone"
    )
    days = [t.date for t in bill.occurrences(dt.date(2026, 3, 31))]

    assert days == [
        dt.date(2026, 1, 31),
        dt.date(2026, 2, 28),
        dt.date(2026, 3, 31),
    ]


def test_report_is_abstract():
    with pytest.raises(TypeError):
        Report()


def test_save_and_load(tmp_path):
    account = Account()
    account.add(tx("2026-01-01", "12.10", "salary"))

    save(account, tmp_path / "b.json")
    loaded = load(tmp_path / "b.json")

    assert loaded.account.transactions == account.transactions


def test_corrupt_file_is_preserved(tmp_path):
    path = tmp_path / "b.json"
    path.write_text("{ not json")

    result = load(path)

    assert result.account.balance == 0
    assert result.problems
    assert (tmp_path / "b.json.corrupt").read_text() == "{ not json"


def test_bad_import_row_reports_line():
    text = (
        "date,description,amount,category\n"
        "2026-01-01,x,5,food\n"
        "2026-01-02,y,oops,food\n"
    )

    with pytest.raises(StatementImportError) as error:
        parse_statement(text)

    assert error.value.line == 3


def test_failed_import_job():
    job = ImportJob()
    job.start(
        "date,description,amount,category\n"
        "2026-01-01,x,oops,food\n"
    )
    job.join(timeout=5)

    assert job.status == "failed"
    assert "line 2" in job.error
    assert job.result is None


class Stop(BaseException):
    pass


@pytest.mark.filterwarnings(
    "ignore::pytest.PytestUnhandledThreadExceptionWarning"
)
def test_worker_stopped_by_baseexception(monkeypatch):
    def die(*args, **kwargs):
        raise Stop()

    monkeypatch.setattr("mybudget.importer.parse_statement", die)

    job = ImportJob()
    job.start("date,description,amount,category\n")
    job.join(timeout=5)

    assert job.status == "failed"
    assert "stopped unexpectedly" in job.error
````
